```python
#count each word in a text
text=input("enter the text seperated by space: ")
words=text.split()
print("text is:",text)
for w in set(words):
    print(w,":",words.count(w))
```

```python
#prompt the user for a list of integers for all values greater than  100 store 'over' instead
numbers = input("Enter integers separated by commas: ")
numbers = [int(x) for x in numbers.split(",")]

for i in range(len(numbers)):
    if numbers[i] > 100:
        numbers[i] = "over"

print("New list:", numbers)
```

```python
#store a list of first names.count the occurences of "a" within list
names=input("Enter first names separated by spaces:").split()
count=0
for n in names:
    count+= n.lower().count('a')
print("Number of occurences:",count)
```

```python
#find gcd of 2 numbers
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
while b !=0:
    a,b=b,a%b
print("GCD=",a)
```

```python
#from alist of integers,create a list removing even numbers
numbers = list(map(int,input("Enter integers:").split()))
result =[]
for num in numbers:
    if num %2 !=0:
        result.append(num)
print("List after removing even numbers:",result)
```

```python
#program to find the factorial of a number
n =int(input("Enter a number:"))
fact = 1
for i in range(1,n+1):
    fact=fact * i
print("Factorial =",fact)
```

```python
#generate fibonacci series of N terms
n=int(input("Enter the number of terms:"))
a = 0
b = 1
for i in range(n):
    print(a,end=" ")
    a,b=b,a+b
```

```python
#find the sum of all items in a list
numbers = list(map(int,input("Enter numbers:").split()))
total = 0
for num in numbers:
    total=total+num
print("Sum =",total)
```

```python
#generate a list of four digit numbers in a givn range with all their digits even and the numbers is a perfect square
import math
start=int(input("Enter starting range:"))
end=int(input("Enter ending range:")) 
result=[]
for num in range(start,end+1):
    if 1000<=num<=9999:
        if int(math.sqrt(num))**2==num:
            if all(int(digit)%2==0 for digit in str(num)):
                result.append(num)
print("Result:",result)
```

```python
#character frequency
text=input("Enter a string:")
count={}
for char in text:
    count[char]=count.get(char,0)+1
print("Character Frequency:")
for char,frequency in count.items():
    print(char," ",frequency)
    
```

```python
#compare two lists
l1=list(map(int,input("Enter first list:").split()))
l2=list(map(int,input("Enter second list:").split()))
#check same length
if len(l1)==len(l2):
    print("Both lists have the same length")
else:
     print("Both lists do not have the same length")
#check same sum
if sum(l1)==sum(l2):
    print("Both lists have the same sum")
else:
    print("Both lists have do not have the same sum")
#check same value
if set(l1)&set(l2):
    print("Both lists have a common value")
else:
    print ("No common values")
```
