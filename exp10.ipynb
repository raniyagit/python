```python
t=input("enter a text:")
words=t.split()
word_counts={}
for word in words:
    if word in word_counts:
        word_counts[word] +=1
    else:
        word_counts[word] =1
print("\nword counts: ")
for word,count in word_counts.items():
    print(f"{word}:{count}")
```

```python
numbers = list(map(int, input("Enter integers: ").split()))

for i in range(len(numbers)):
    if numbers[i] > 100:
        numbers[i] = "over"

print(numbers)
```
