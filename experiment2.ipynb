```python
#change the string
string=input("enter the string: ")
if len(string)<=1:
    print(string)
else:
    new_string=string[-1]+string[1:-1]+string[0]
    print("new string is: ",new_string)
    
```

```python
#string made of 1st and last 2 charecters
string=input("enter a string: ")
if len(string)<2:
    print("string is too short")
else:
    new_string=string[:2]+string[-2:]
    print("new string is: ",new_string)
```

```python
#string sep with space by swapping the charecter
s1=input("enter first string: ")
s2=input("enter second string: ")
new_s1=s2[0]+s1[1:]
new_s2=s1[0]+s2[1:]
print(new_s1,new_s2)
```

```python
#string from given string($)
string=input("enter a string: ")
first_char=string[0]
new_string=first_char+string[0]
new_string=first_char+string[1:].replace(first_char,"$")
print("new string is: ",new_string)
```
