```python
#display first and last colors
a=input("enter the colors: ")
color=a.split(",")
print("first color: ",color[0])
print("second color: ",color[-1])
```

```python
#extension
filename=input("enter ths filename ")
extension=filename.split(".")[-1]
print("extension: ",extension)
```

```python
#sum of numbers
numbers=input("enter comma-seperated numbers :")
num_list=numbers.split(",")
num_list=[int(x) for x in num_list]
print("list: ",num_list)
print("sum: ",sum(num_list))
```

```python
#3rd element
numbers=(10,20,30,40,50)
print("3rd elements: ",numbers[2])
print("3rd element from last: ",numbers[-3])
```

```python
#commas
numbers=input("enter numbers seperated by commas: ")
num_tuple=tuple(map(float,numbers.split(",")))
average=sum(num_tuple)/len(num_tuple)
print("tuple:",num_tuple)
print("average:",average)
```
