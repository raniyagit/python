```python
try:
    age = int(input("Enter your age: "))
    if age <= 0:
        raise ValueError("Age must be a positive number.")
        print("Your age is:", age)
except ValueError as e:
    print("Invalid input!", e)
finally:
    print("Program execution completed.")
```
