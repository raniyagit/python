```python
numbers=input("Enter the numbers separated by commas:").split(",")
positive_num= [int(n) for n in numbers if int(n) >0]
print(positive_num)
```

```python
n=int(input("Enter the N number:"))
print("The number N is",n)
squares= [x**2 for x in range(1, n + 1)]
print(squares)
```

```python
word=input("Enter a word:")
print("The word is",word)
vowels=[char for char in word if char.lower() in "a,e,i,o,u"]
print(vowels)
```

```python
word=input("Enter the word:")
print("The word is:",word)
ordinal=[ ord(char) for char in word]
print(ordinal)
```
